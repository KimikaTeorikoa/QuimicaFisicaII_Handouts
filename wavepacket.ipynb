# Construcción de un paquete de ondas
Aquí vamos a ilustrar cómo a partir de una función trigonométrica sencilla podemos construir, mediante superposición de un conjunto de ellas, un paquete de ondas.

En primer lugar importamos una serie de librerías de Python

```
import numpy as np
import matplotlib.pyplot as plt
```

A continuación, generamos un vector que va de -$\pi$ a $\pi$

```
x = np.linspace(-2*np.pi, 2*np.pi, 1000)
```

Finalmente escribimos una función de Python, que nos devuelve una función que suma el coseno y el seno.

```
def sin_cos(x, k=1):
  return np.cos(k*x) - np.sin(k*x)*1j
```

```
fig, ax = plt.subplots(2,1, sharex=True)
psi = sin_cos(x)
ax[0].plot(x, psi, lw=3)
ax[0].plot(x, np.imag(psi), lw=3)
ax[1].plot(x, np.conj(psi)*psi, lw=3)
ax[0].set_ylabel(r'$\psi(x)$', fontsize=14)
ax[1].set_ylabel(r'$\psi^2(x)$', fontsize=14)
ax[1].set_xlabel(r'$x$', fontsize=14)
ax[1].set_xlim(x[0], x[-1])
plt.tight_layout()
```

Sin embargo, si superponemos un número de funciones con diferentes periodos, veremos que la densidad se acumula en una región en particular.

```
fig, ax = plt.subplots(2,1, sharex=True)
psi = np.zeros(len(x), dtype=complex)
for k in range(1,50):
  psi += sin_cos(x, k)
ax[0].plot(x, psi, lw=3)
ax[1].plot(x, np.conj(psi)*psi, lw=3)
ax[0].set_ylabel(r'$\psi(x)$', fontsize=12)
ax[1].set_ylabel(r'$\psi^2(x)$', fontsize=12)
ax[1].set_xlabel(r'$x$', fontsize=12)
ax[1].set_xlim(-np.pi, np.pi)
plt.tight_layout()
```
