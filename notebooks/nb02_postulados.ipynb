```python
import numpy as np
import matplotlib.pyplot as plt
import qf2figs as qf
```

## 1. Funciones de onda aceptables y no aceptables

```python
fig, axes = plt.subplots(2, 2, figsize=(qf.TEXT, 3.3), sharex=True)
x = np.linspace(-3, 3, 800)

# (a) multivaluada
ax = axes[0, 0]
t = np.linspace(-np.pi / 2 + 0.05, np.pi / 2 - 0.05, 400)
ax.plot(1.6 * np.sin(2 * t), 1.3 * np.sin(t), color=qf.PALETTE[1])
ax.set_title("(a) no unívoca", loc="left", fontsize=8, color=qf.PALETTE[1])

# (b) discontinua
ax = axes[0, 1]
xa, xb = x[x < 0], x[x >= 0]
ax.plot(xa, np.exp(-xa**2) * 0.5, color=qf.PALETTE[1])
ax.plot(xb, np.exp(-xb**2) * 1.2 - 0.35, color=qf.PALETTE[1])
ax.plot(0, 0.5, "o", mfc="white", mec=qf.PALETTE[1], ms=4, mew=1.0)
ax.plot(0, 0.85, "o", color=qf.PALETTE[1], ms=4)
ax.set_title("(b) discontinua", loc="left", fontsize=8, color=qf.PALETTE[1])

# (c) derivada discontinua
ax = axes[1, 0]
ax.plot(x, np.exp(-np.abs(x) * 1.6), color=qf.PALETTE[1])
ax.set_title("(c) derivada discontinua", loc="left", fontsize=8,
             color=qf.PALETTE[1])

# (d) aceptable
ax = axes[1, 1]
ax.plot(x, 1.15 * x * np.exp(-x**2), color=qf.PALETTE[2])
ax.set_title("(d) aceptable", loc="left", fontsize=8, color=qf.PALETTE[2])

for ax in axes.flat:
    ax.axhline(0, color=qf.MUTED, lw=0.5)
    ax.set_yticks([])
    ax.set_xticks([])
    ax.spines["left"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.set_xlim(-3, 3)
fig.supxlabel(r"$x$", fontsize=9, y=0.02)
fig.suptitle("Requisitos de una función de onda admisible", fontsize=9,
             x=0.02, ha="left")
fig.tight_layout(rect=(0, 0.02, 1, 0.94))
qf.save(fig, "postulados", "funciones-admisibles")
```

## 2. Paquete de ondas: localización por superposición

```python
fig, axes = plt.subplots(1, 4, figsize=(qf.FULL, 1.9), sharey=True)
x = np.linspace(-12, 12, 2000)
k0, dk = 3.0, 0.55
for ax, N in zip(axes, [1, 3, 11, 81]):
    ks = np.array([k0]) if N == 1 else np.linspace(k0 - 2.2, k0 + 2.2, N)
    w = np.exp(-((ks - k0) ** 2) / (2 * dk ** 2))
    psi = (w[:, None] * np.cos(ks[:, None] * x)).sum(0) / w.sum()
    ax.plot(x, psi, color=qf.PALETTE[0], lw=1.0)
    ax.set_title(f"$N={N}$", fontsize=8, loc="left")
    ax.axhline(0, color=qf.MUTED, lw=0.5)
    ax.set_yticks([])
    ax.set_xticks([-10, 0, 10])
    ax.spines["left"].set_visible(False)
    ax.set_xlabel(r"$x$", fontsize=8)
axes[0].set_ylabel(r"$\Psi(x)$", fontsize=8)
fig.suptitle(r"Superposición de $N$ ondas armónicas: al localizar en $x$ "
             r"se pierde la definición de $p$", fontsize=8.5, x=0.02,
             ha="left")
fig.tight_layout(rect=(0, 0, 1, 0.90))
qf.save(fig, "incertidumbre", "paquete-ondas")
```

## 3. Par x / k: el principio de incertidumbre

```python
fig, axes = plt.subplots(1, 2, figsize=(qf.TEXT, 2.1))
x = np.linspace(-8, 8, 1000)
p_mom = np.linspace(-8, 8, 1000)   # eje de momentos (p = hbar k)
# Alturas escalonadas: las tres curvas valen 1 en el máximo, así que si se
# etiquetan todas cerca de x=0 las tres etiquetas caen encima unas de otras.
# Se coloca cada una a una altura distinta de su propia curva, despejando
# x = sigma*sqrt(-2 ln y).
for i, (s, y_lab) in enumerate(zip([0.5, 1.2, 2.6], [0.72, 0.52, 0.32])):
    axes[0].plot(x, np.exp(-x**2 / (2 * s**2)), color=qf.PALETTE[i])
    axes[1].plot(p_mom, np.exp(-p_mom**2 * s**2 / 2), color=qf.PALETTE[i])
    x_lab = s * np.sqrt(-2 * np.log(y_lab))
    axes[0].text(x_lab + 0.25, y_lab, rf"$\sigma={s}$", color=qf.PALETTE[i],
                 fontsize=7, ha="left", va="bottom")
axes[0].set_xlabel(r"$x$")
axes[0].set_ylabel(r"$|\psi(x)|$")
axes[0].set_title(r"espacio de posiciones", loc="left", fontsize=8)
axes[1].set_xlabel(r"$p$")
axes[1].set_ylabel(r"$|\tilde\psi(p)|$")
axes[1].set_title(r"espacio de momentos", loc="left", fontsize=8)
for ax in axes:
    ax.set_yticks([])
    ax.spines["left"].set_visible(False)
fig.suptitle(r"Estrechar en $x$ ensancha en $p$:  $\Delta x\,\Delta p\geq \hbar/2$",
             fontsize=8.5, x=0.02, ha="left")
fig.tight_layout(rect=(0, 0, 1, 0.90))
qf.save(fig, "incertidumbre", "x-p")
```

## 4. Función de onda y densidad de probabilidad

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(qf.TEXT, 3.1), sharex=True)

# Paquete gaussiano modulado: a diferencia de una gaussiana pelada, cambia de
# signo, que es justo lo que hace visible el efecto de elevar al cuadrado.
x = np.linspace(-7, 7, 2000)
sigma, k0 = 2.0, 1.0
psi = np.exp(-x**2 / (4 * sigma**2)) * np.cos(k0 * x)
psi = psi / np.sqrt(np.trapz(psi**2, x))          # normalizada
dens = psi**2

# Nodos: donde cos(k0 x) = 0
nodos = np.array([-3, -1, 1, 3]) * np.pi / (2 * k0)

# --- arriba: psi, que toma valores negativos ---
ax1.fill_between(x, psi, 0, where=psi >= 0, color=qf.FILL_POS, alpha=0.22, lw=0)
ax1.fill_between(x, psi, 0, where=psi < 0, color=qf.FILL_NEG, alpha=0.22, lw=0)
ax1.plot(x, psi, color=qf.PALETTE[0])
ax1.set_ylabel(r"$\psi(x)$")
ax1.text(-2.83, psi.min() * 0.50, r"$\psi<0$", color=qf.FILL_NEG, fontsize=8,
         ha="center", va="center")

# --- abajo: |psi|^2, nunca negativa, y cero en los mismos nodos ---
ax2.fill_between(x, dens, 0, color=qf.PALETTE[1], alpha=0.22, lw=0)
ax2.plot(x, dens, color=qf.PALETTE[1])
ax2.set_ylabel(r"$|\psi(x)|^2$")
ax2.annotate("nodos", xy=(nodos[2], 0.0), xytext=(nodos[2], dens.max() * 0.70),
             fontsize=8, color=qf.ACCENT, ha="center", va="bottom",
             arrowprops=dict(arrowstyle="->", color=qf.ACCENT, lw=0.8))

for ax in (ax1, ax2):
    qf.zero_axis(ax)
    for xn in nodos:
        ax.axvline(xn, color=qf.MUTED, lw=0.5, ls=":", zorder=0)
    ax.set_yticks([])
    ax.spines["left"].set_visible(False)
    ax.set_xlim(-7, 7)

# el spine inferior del panel de arriba se leería como un segundo cero
ax1.spines["bottom"].set_visible(False)
ax1.tick_params(bottom=False)

ax2.set_xlabel(r"$x$")
fig.tight_layout()
qf.save(fig, "postulados", "psi-psi2")

# Comprobaciones numéricas (tests de regresión)
print(f"  norma        = {np.trapz(dens, x):.3f}   (debe ser 1.000)")
print(f"  min(psi)     = {psi.min():.3f}   (debe ser < 0)")
print(f"  min(|psi|^2) = {dens.min():.3f}   (debe ser >= 0)")
print(f"  nodos en x   = {np.round(nodos, 3)}")
```
